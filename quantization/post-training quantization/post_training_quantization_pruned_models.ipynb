```python
DATA_YAML = "/content/gdrive/MyDrive/dataset/dataset/data.yaml"

#Customizedd model
simple_customized_yolov8_pruned = "/content/gdrive/MyDrive/EE_using_ultralitycs/results/Model_without_EE/pruned/customized_without_EE_pruned.pt"

#Early exit - BackBone neck
best_EE_Backbone_Neck_pruned= "/content/gdrive/MyDrive/EE_using_ultralitycs/results/EE_Backbone_Neck/pruned/EE_backbone_neck_pruned.pt"

#Early exit - EE_Backboneonly
best_EE_Backboneonly_pruned= "/content/gdrive/MyDrive/EE_using_ultralitycs/results/EE_Backboneonly/pruned/EE_backboneonly_pruned.pt"
```

#**EE - Backbone and Neck**

## post-training full integer (INT8) quantization

```python
from ultralytics import YOLO

model = YOLO(best_EE_Backbone_Neck_pruned)

model.export(
    data = DATA_YAML,
    format="tflite",
    int8=True,
    device="cuda"
)
```

## post-training Float16 quantization



```python
from ultralytics import YOLO

model = YOLO(best_EE_Backbone_Neck_pruned)

model.export(
    format="tflite",
    half=True,
    device=0
)
```
