#**EE - Backbone and Neck**

## post-training full integer (INT8) quantization

```python
from ultralytics import YOLO

model = YOLO(best_EE_Backbone_Neck)

model.export(
    data = DATA_YAML,
    format="tflite",
    int8=True,
    device="cuda"
)
```

## post-training Float16 quantization



```python
model = YOLO(best_EE_Backbone_Neck)

model.export(
    format="tflite",
    half=True,
    device=0
)
```
